You are a data analyst working with a university's placement cell. They want to understand *what factors influence whether a student gets placed*, and use that insight to advise current students and design intervention programs for at-risk students.

You've been given a dataset of 100,000 student records containing academic performance, skills, extracurricular activity, lifestyle habits, and placement outcomes. Your job is to clean the data, explore it, and deliver actionable findings.

##  Part A: Data Loading & Cleaning (20 points)

```python
import pandas as pd 
import numpy as np 
import matplotlib.pyplot as plt 
import seaborn as sns
```

### 1. Load the dataset and report its shape, column data types, and memory usage.

```python
# Loading Dataset
student_df = pd.read_csv('student.csv')
student_df.head()
```

```python
# Shape of the Data 
print("Shape :",student_df.shape)
```

```python
# Column Datatypes 
print('Column Datatypes')
for i in student_df.columns: 
    print(f'{i} : {student_df[i].dtype}')
```

```python
# Memory Usage
student_df.info()
```

### 2. Identify and report all columns containing missing or suspicious values (e.g., extracurricular_score has some 0 values that look like placeholders rather than genuine zeros — investigate and decide how to handle them, with justification).

```python
# missing or suspicious values 
print("N/A Values:")
print(student_df.isna().sum())
print("\n\nNull Values:") 
student_df.isnull().sum()
```

```python
# Finding all vlaues where values is zero
(student_df.select_dtypes(include='number') == 0).sum()
```

```python
# dropping extracurricular_score as in some fields, students have extracurricular activities but 0 extracurricular
student_df = student_df[student_df['extracurricular_score'] != 0]
student_df
```

### 3. Check for and remove any duplicate student_id entries.

```python
student_df['student_id'].duplicated().sum()
```

```python
student_df = student_df.drop_duplicates(subset=['student_id'], keep='first') 
student_df.shape
```

### 4. Drop or fix the Checking column (a stray boolean column with no clear purpose) — document your reasoning either way.

```python
# Dropping the column Checking as it does not have a clear purpose or meaning
student_df.drop('Checking', inplace=True, axis=1)
student_df.head()
```

### 5. Convert placement_status into a numeric column placed_flag (1/0) for later correlation analysis.


```python
# Converting Placement status to numeric columns
student_df['Placement Flag'] = student_df['placement_status'].map({'Placed':1,'Not Placed':0})
student_df[['placement_status', 'Placement Flag']].head(20)
```

## Part C: Feature Engineering (20 points)

### 12. Create a total_experience_score column combining internships_count, projects_count, certifications_count, and hackathons_participated (choose a weighting scheme and justify it).

```python
student_df['total_experience_score'] = student_df[['internships_count','projects_count','certifications_count','hackathons_participated']].sum(axis=1)
student_df[['internships_count','projects_count','certifications_count','hackathons_participated','total_experience_score']].head(10)
```

### 13. Bin study_hours_per_day into Low / Medium / High categories using pd.cut(), and compare placement rates across bins.

```python
student_df['study_quality'] = pd.cut(student_df['study_hours_per_day'],3,labels=['bad','medium','good']) 
student_df[['study_quality','study_hours_per_day']].head(10)
```

### 14. Create a well_rounded flag: True if a student has leadership_score > 60, extracurricular_score > 60, AND cgpa > 7.5. What fraction of students qualify, and what's their placement rate compared to the rest?

```python
# creating the well rounded column where 1 indicates True and 0 indicates False
student_df['well_rounded'] = np.where((student_df['leadership_score']>60) & (student_df['cgpa']>7.5),1,0)
student_df.head(20)
```

```python
# The fraction of students who qualify for being well rounded students 
print('Percent of well rounded students:',student_df['well_rounded'].mean()*100)
```

```python
# Placement rate of students depending on their well rounded score
student_df[['well_rounded', 'Placement Flag']].groupby('well_rounded').mean()
```

### 15. Create an at_risk flag for students with backlogs >= 2 and attendance_percentage < 75. How many such students exist, and what's their placement rate? 

```python
# creating the at risk flag
student_df['at_risk'] = np.where((student_df['attendance_percentage'] < 74) & (student_df['backlogs'] >= 2),1,0)
student_df.head(20)
```

```python
# no of at risk students 
print('No of students at risk are:',student_df['at_risk'][student_df['at_risk'] == 1].sum())
```

```python
# Placement rate of students depending on their risk score score
student_df[['at_risk', 'Placement Flag']].groupby('at_risk').mean()
```

## Part D: Aggregation & Business Insight (20 points)

### 16. Build a .pivot_table() showing average salary_package_lpa with branch as rows and college_tier as columns.

```python
pd.pivot_table(student_df,index='college_tier',values='salary_package_lpa',aggfunc='mean')
```

### 17. Using .groupby() + .agg(), produce a single summary table per branch showing: student count, placement rate, average CGPA, and average salary — sorted by placement rate descending.

```python
# branch wise total summary 
student_df.groupby('branch').agg({'student_id':'count', 'Placement Flag': 'mean', 'cgpa': 'mean', 'salary_package_lpa': 'mean'}).sort_values('Placement Flag', ascending=False)
```

### 18. Identify the *top 5%* of students by salary_package_lpa among placed students. What do they have in common (skills, branch, tier)?

```python
top5 = student_df[student_df['Placement Flag'] == 1][student_df["salary_package_lpa"]>student_df["salary_package_lpa"].quantile(0.95)]
top5.head()
```

```python
# common branch, skills and other parameters
top5[[ 'branch', 'college_tier','total_experience_score', 'sleep_hours', 'study_quality','extracurricular_score','salary_package_lpa']].value_counts(normalize=True).head()
```
